# 00. Kiểm tra môi trường Jupyter

**Mục tiêu:** xác nhận Python và các thư viện đã sẵn sàng trước khi chạy thực nghiệm.

**Cách làm:** Mở notebook này trong Jupyter → **Kernel > Restart Kernel and Run All Cells** (hoặc bấm `Shift+Enter` từng ô). Nếu một thư viện báo **CHƯA CÀI**, hãy xem `HUONG_DAN_CHAY.md`.

```python
import sys
import importlib
from pathlib import Path

print('Đường dẫn Python:', sys.executable)
print('Phiên bản Python:', sys.version.split()[0])
print('Thư mục đang làm việc:', Path.cwd())

packages = ['numpy', 'pandas', 'matplotlib', 'sklearn', 'torch', 'nbformat']
for name in packages:
    try:
        module = importlib.import_module(name)
        print(f'{name:12}: OK – {getattr(module, "__version__", "đã cài")}')
    except ImportError:
        print(f'{name:12}: CHƯA CÀI')
```

**Nếu thiếu thư viện:** Trong Anaconda Prompt/Terminal chạy `python -m pip install -r requirements.txt` tại thư mục chứa các notebook, sau đó khởi động lại Kernel. Không cần tải file dữ liệu riêng: bộ ảnh chữ số đi kèm `scikit-learn`, chuỗi nhiệt độ được tạo bằng mã và được **gắn nhãn mô phỏng**.
